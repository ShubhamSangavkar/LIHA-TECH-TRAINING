## **FUNCTION WITH RETURN**

```python
def test(temp):
    if temp > 90:
        return '+ve'
    else:
        return '-ve'

result = test(89)

if result == "+ve":
    print("Quaranteen")
else:
    print("Home Isolation")
```

# **Higher order function**

```python
# A Function who needs another function as an input called as higher order function
# map(fun, iterable)
# filter(fun,iterable)
# reduce(fun,iterable)
```

**MAP : map(function , iterable)**

```python
data = ['male', 'female', 'female', 'male']

convert = lambda gen: 0 if gen=="male" else 1

list(map(convert,data))
```

**FILTER : filter(function,iterable)**

```python
nums=[10,20,30,40]

list(filter(lambda val:val>25 , nums))
```

```python
list(filter(lambda x: x%5 == 0 or x%7 == 0 , range(1,31)))
```

**REDUCE : reduce(function,iterable)**

```python
nums
```

```python
from functools import reduce 
```

```python
reduce(lambda x,y:max(x,y) , nums)
```
