📌 Problem 1: Remove Duplicates from a List

Given a list of integers, create a new list containing only unique elements while preserving their original order.

Input:

numbers = [4, 2, 4, 7, 2, 9, 7, 1]

Expected Output:

[4, 2, 7, 9, 1]

Requirements:

- Do not modify the original list.
- Do not use a set to remove duplicates.
- Handle empty lists and lists containing only duplicate values.
- Bonus: Analyze the time and space complexity of your solution.

```python
numbers = [4, 2, 4, 7, 2, 9, 7, 1]
res=[]
for i in numbers:
    if i not in res:
        res.append(i)
print(res)
```


📌 Problem 2: Word Frequency Counter

Write a program to count how many times each word appears in a sentence, ignoring capitalization.

Input:

sentence = "Python is great and Python is powerful"

Expected Output:

{
    "python": 2,
    "is": 2,
    "great": 1,
    "and": 1,
    "powerful": 1
}

Requirements:

- Use a dictionary to store word frequencies.
- Handle extra spaces between words.
- Identify the most frequently occurring word.


```python
sentence = "Python is great and Python is powerful"
res = {}
words=sentence.split()
for i in words:
    if i in res.keys():
        res[i]+=1
    else:
        res[i]=1
print(res)
```


📌 Problem 3: Find the Second-Largest Distinct Number

Given an unsorted list of integers, find the second-largest distinct number without sorting the list.

Input:

numbers = [10, 5, 20, 20, 8, 15]

Expected Output:

15

Requirements:

- Do not use "sort()" or "sorted()".
- Handle duplicate maximum values correctly.
- Decide how to handle lists with fewer than two distinct numbers.
- Bonus: Target O(n) time complexity.


```python
numbers = [10, 5, 20, 20, 8, 15]
largest = None
second = None
for n in numbers:
    if largest is None:
        largest = n
    elif n > largest:
        second = largest
        largest = n
    elif n < largest and (second is None or n > second):
        second = n
        
print(second)
```

]

{

}

Requirements:

- Use a loop and dictionary.
- Support any number of employees and departments.
- Preserve employee order within each department.
- Bonus: Explain how this pattern can be applied to grouping ML experiment results.


```python
employees = [
    {"name": "Amit", "dept": "AI"},
    {"name": "Priya", "dept": "Data"},
    {"name": "Rahul", "dept": "AI"},
    {"name": "Neha", "dept": "Data"},
    {"name": "John", "dept": "Cloud"}
]

res={}
for i in employees:
    if i['dept'] in res.keys():
        res[i['dept']].append(i['name'])
    else:
        res[i['dept']]=[i['name']]
res 
```

📌 Problem 6: Longest Consecutive Sequence

Given an unsorted list of integers, find the length of the longest sequence of consecutive integers. The numbers do not need to appear next to each other in the original list.

Input:

numbers = [100, 4, 200, 1, 3, 2, 5]

Expected Output:

5

The longest consecutive sequence is "1, 2, 3, 4, 5".

Requirements:

- Do not sort the list.
- Target O(n) average time complexity.
- Handle duplicates, negative numbers, and empty lists.
- Bonus: Explain the time and space complexity of your approach.



```python
numbers = [100, 4, 200, 1, 3, 2, 5]
num_set = numbers
longest = 0
for n in num_set:
    if n - 1 not in num_set:
        current = n
        length = 1
        while current + 1 in num_set:
            current += 1
            length += 1
        if length > longest:
            longest = length
longest
            
```
