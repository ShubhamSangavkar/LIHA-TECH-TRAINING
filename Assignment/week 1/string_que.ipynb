```python
s = "aaabbccccd"
result = ""
i = 0
while i < len(s):
    j = i
    while j < len(s) and s[j] == s[i]:
        j += 1
    result += s[i] + str(j-i)
    i = j
print("Compressed string =", result)
```

 Q27


```python
s = "abcabcbb"
seen = set()
left = 0
best = ""
for right, ch in enumerate(s):
    while ch in seen:
        seen.remove(s[left])
        left += 1
    seen.add(ch)
    if right-left+1 > len(best):
        best = s[left:right+1]
print("Longest substring =", best)
print("Length =", len(best))
```

Q28


```python
s = "babad"
best = ""
for i in range(len(s)):
    for j in range(i, len(s)):
        x = s[i:j+1]
        if x == x[::-1] and len(x) > len(best):
            best = x
print("One longest palindromic substring =", best)
```

Q29


```python
text = "Hello World"
shift = 3
result = ""
for ch in text:
    if ch.isalpha():
        base = ord('A') if ch.isupper() else ord('a')
        result += chr((ord(ch)-base+shift)%26+base)
    else:
        result += ch
print("Encrypted text =", result)
```

Q30


```python
s = "a3b2c4"
result = ""
i = 0
while i < len(s):
    ch = s[i]
    i += 1
    num = ""
    while i < len(s) and s[i].isdigit():
        num += s[i]
        i += 1
    result += ch * int(num)
print("Decoded string =", result)
```

Q31


```python
s = "{[()()]}"
pairs = {')':'(', ']':'[', '}':'{'}
stack = []
for ch in s:
    if ch in "([{":
        stack.append(ch)
    elif not stack or stack.pop() != pairs[ch]:
        print("Balanced =", False)
        break
else:
    print("Balanced =", not stack)
```

Q36


```python
s = "ADOBECODEBANC"
pattern = "ABC"
need = {c: pattern.count(c) for c in set(pattern)}
best = None
for i in range(len(s)):
    count = {}
    for j in range(i, len(s)):
        count[s[j]] = count.get(s[j], 0) + 1
        if all(count.get(c, 0) >= n for c, n in need.items()):
            window = s[i:j+1]
            if best is None or len(window) < len(best):
                best = window
            break
print("Minimum window =", best)
```
