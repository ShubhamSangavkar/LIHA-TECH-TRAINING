# Python Practice Sheet - Lambda Function

## Section A - Basic

**Q1.** Lambda function to return the square of a number.

```python
square = lambda x: x * x
print(square(9))
```

**Q2.** Lambda function to return the cube of a number.

```python
cube = lambda x: x ** 3
print(cube(4))
```

**Q3.** Lambda function to add two numbers.

```python
add = lambda a, b: a + b
print(add(15, 25))
```

**Q4.** Lambda function to return the greater of two numbers.

```python
greater = lambda a, b: a if a > b else b
print(greater(18, 27))
```

**Q5.** Lambda function to check whether a number is even.

```python
is_even = lambda x: x % 2 == 0
print(is_even(14))
```

**Q6.** Lambda function to return Positive, Negative, or Zero based on a number.

```python
sign = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(sign(-8))
```

**Q7.** Lambda function to return the first character of a string.

```python
first_char = lambda text: text[0]
print(first_char("Python"))
```

**Q8.** Lambda function to return the length of a string.

```python
string_length = lambda text: len(text)
print(string_length("LihaTech"))
```

**Q9.** Lambda function to convert a string to uppercase.

```python
to_upper = lambda text: text.upper()
print(to_upper("python"))
```

**Q10.** Lambda function to multiply three numbers.

```python
multiply = lambda a, b, c: a * b * c
print(multiply(2, 3, 4))
```

**Q11.** Sort a list of tuples by the second value using a lambda function.

```python
items = [(1, 30), (2, 10), (3, 20)]
result = sorted(items, key=lambda t: t[1])
print(result)
```

**Q12.** Sort words by their length using a lambda function.

```python
words = ['python', 'AI', 'data', 'cloud']
result = sorted(words, key=lambda w: len(w))
print(result)
```

**Q13.** Square all numbers in a list using map() and lambda.

```python
numbers = [1, 2, 3, 4, 5]
result = list(map(lambda x: x * x, numbers))
print(result)
```

**Q14.** Select all even numbers using filter() and lambda.

```python
numbers = [1, 2, 3, 4, 5, 6]
result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)
```

**Q15.** Select words beginning with A using filter() and lambda.

```python
words = ['Apple', 'Python', 'AI', 'Data', 'Agent']
result = list(filter(lambda w: w.startswith('A'), words))
print(result)
```

**Q16.** Convert Celsius values to Fahrenheit using map() and lambda.

```python
celsius = [0, 20, 30, 40]
fahrenheit = list(map(lambda c: c * 9 / 5 + 32, celsius))
print(fahrenheit)
```

## Section B - Intermediate

**Q1.** Sort a list of employee dictionaries by age using a lambda key.

```python
employees = [{'name': 'Asha', 'age': 29}, {'name': 'Ravi', 'age': 24}, {'name': 'Neha', 'age': 31}]
sorted_employees = sorted(employees, key=lambda e: e['age'])
names = [e['name'] for e in sorted_employees]
print(names)
```

**Q2.** Sort a list of names by the last character using a lambda function.

```python
names = ['Asha', 'Ravi', 'Neha', 'Aman']
result = sorted(names, key=lambda n: n[-1])
print(result)
```

**Q3.** Apply a 10% discount to every price using map() and lambda.

```python
prices = [500, 1000, 1500, 2000]
discounted = list(map(lambda p: p - (p * 10 / 100), prices))
print(discounted)
```

**Q4.** Filter student records with marks of 75 or above using filter() and lambda.

```python
students = [('Asha', 82), ('Ravi', 68), ('Neha', 91), ('Aman', 75)]
result = list(filter(lambda s: s[1] >= 75, students))
print(result)
```

**Q5.** Calculate the product of all numbers using reduce() and lambda.

```python
from functools import reduce

numbers = [2, 3, 4, 5]
product = reduce(lambda a, b: a * b, numbers)
print("Product =", product)
```

**Q6.** Filter palindrome words using lambda.

```python
words = ['level', 'python', 'madam', 'cloud', 'radar']
result = list(filter(lambda w: w == w[::-1], words))
print(result)
```

**Q7.** Transform each number into a tuple containing the number, square, and cube using map() and lambda.

```python
numbers = [1, 2, 3, 4]
result = list(map(lambda n: (n, n ** 2, n ** 3), numbers))
print(result)
```

**Q8.** Sort strings without considering uppercase or lowercase using a lambda key.

```python
words = ['banana', 'Apple', 'cherry', 'apricot']
result = sorted(words, key=lambda w: w.lower())
print(result)
```

**Q9.** Find the dictionary key having the maximum value using a lambda key.

```python
scores = {'Asha': 82, 'Ravi': 74, 'Neha': 91}
top = max(scores, key=lambda k: scores[k])
print(top)
```

**Q10.** Sort student records by marks in descending order and then by name in ascending order using lambda.

```python
students = [('Ravi', 85), ('Asha', 92), ('Neha', 85), ('Aman', 92)]
result = sorted(students, key=lambda s: (-s[1], s[0]))
print(result)
```

## Section C - Advanced

**Q1.** Lambda expression to assign grades: A for 90+, B for 75-89, C for 60-74, and D below 60.

```python
grade = lambda m: 'A' if m >= 90 else 'B' if m >= 75 else 'C' if m >= 60 else 'D'
marks = [95, 81, 69, 52]
print([grade(m) for m in marks])
```

**Q2.** Lambda expression to calculate tax using slabs: 0% up to 300000, 5% up to 600000, and 10% above 600000.

```python
tax = lambda income: 0 if income <= 300000 else (income - 300000) * 0.05 if income <= 600000 else 15000 + (income - 600000) * 0.10
print("Tax =", float(tax(800000)))
```

**Q3.** Use filter(), map(), and lambda to return the squares of only even numbers.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = filter(lambda x: x % 2 == 0, numbers)
result = list(map(lambda x: x * x, evens))
print(result)
```

**Q4.** Sort file-version tuples by major, minor, and patch numbers using a lambda key.

```python
versions = [('app', '1.2.10'), ('app', '1.10.2'), ('app', '1.2.3')]
result = sorted(versions, key=lambda v: tuple(map(int, v[1].split('.'))))
print(result)
```

**Q5.** Sort points by their Manhattan distance from the origin using a lambda key.

```python
points = [(3, 4), (1, 1), (-2, 2), (0, 5)]
result = sorted(points, key=lambda p: abs(p[0]) + abs(p[1]))
print(result)
```

**Q6.** Sort ISO date strings chronologically using a lambda key that converts date parts to integers.

```python
dates = ['2026-12-01', '2025-05-10', '2026-01-15']
result = sorted(dates, key=lambda d: tuple(map(int, d.split('-'))))
print(result)
```

**Q7.** Find the longest word using max() with a lambda key.

```python
words = ['AI', 'Python', 'MachineLearning', 'Data']
longest = max(words, key=lambda w: len(w))
print(longest)
```

**Q8.** Create a calculator dictionary whose values are lambda functions.

```python
calculator = {
    'add': lambda a, b: a + b,
    'subtract': lambda a, b: a - b,
    'multiply': lambda a, b: a * b,
    'divide': lambda a, b: a / b,
}

operation = 'multiply'
a = 7
b = 6
print("Result =", calculator[operation](a, b))
```

**Q9.** Sort transaction records by amount descending and, for equal amounts, by transaction ID ascending using lambda.

```python
transactions = [('T3', 500), ('T1', 700), ('T2', 700), ('T4', 300)]
result = sorted(transactions, key=lambda t: (-t[1], t[0]))
print(result)
```

**Q10.** Use reduce() and lambda to find the largest number without using max().

```python
from functools import reduce

numbers = [12, 45, 7, 91, 33]
largest = reduce(lambda a, b: a if a > b else b, numbers)
print("Largest =", largest)
```
