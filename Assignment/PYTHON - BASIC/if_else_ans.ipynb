```python
# Q10
amount = 2860
notes = [500, 200, 100, 50, 20, 10]
total_notes = 0

for note in notes:
    count = amount // note
    print(note, ":", count)
    total_notes += count
    amount = amount % note

print("Total number of notes =", total_notes)
```

```python
# Q14
actual_cost = 1000
selling_cost = 1200

if selling_cost > actual_cost:
    profit = selling_cost - actual_cost
    print("Profit =", profit)
elif actual_cost > selling_cost:
    loss = actual_cost - selling_cost
    print("Loss =", loss)
else:
    print("No Profit No Loss")
```

```python
# Q15
physics = 85
chemistry = 80
biology = 75
mathematics = 90
computer = 95

total = physics + chemistry + biology + mathematics + computer
percentage = total / 5

print("Percentage =", percentage)

if percentage >= 90:
    print("Grade A")
elif percentage >= 80:
    print("Grade B")
elif percentage >= 70:
    print("Grade C")
elif percentage >= 60:
    print("Grade D")
elif percentage >= 40:
    print("Grade E")
else:
    print("Grade F")
```

```python
# Q16
basic_salary = 15000

if basic_salary <= 10000:
    hra = basic_salary * 20 / 100
    da = basic_salary * 80 / 100
elif basic_salary <= 20000:
    hra = basic_salary * 30 / 100
    da = basic_salary * 90 / 100
else:
    hra = basic_salary * 35 / 100
    da = basic_salary * 95 / 100

gross_salary = basic_salary + hra + da

print("Basic Salary =", basic_salary)
print("HRA =", hra)
print("DA =", da)
print("Gross Salary =", gross_salary)
```

```python
# Q17
units = 275

if units <= 50:
    bill = units * 0.50
elif units <= 150:
    bill = 50 * 0.50 + (units - 50) * 0.75
elif units <= 250:
    bill = 50 * 0.50 + 100 * 0.75 + (units - 150) * 1.25
else:
    bill = 50 * 0.50 + 100 * 0.75 + 100 * 1.25 + (units - 250) * 1.50

surcharge = bill * 17 / 100
total_bill = bill + surcharge

print("Bill =", bill)
print("Surcharge =", surcharge)
print("Total Electricity Bill =", total_bill)
```
