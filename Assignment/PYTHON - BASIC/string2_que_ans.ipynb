03

```python
s="Hello World"
v=c=0
for ch in s.lower():
    if ch in "aeiou": v+=1
    elif ch.isalpha(): c+=1
print("Vowels =",v)
print("Consonants =",c)
```

04

```python
s="Hello@123 World!"
u=l=d=sp=0
for ch in s:
    if ch.isupper(): u+=1
    elif ch.islower(): l+=1
    elif ch.isdigit(): d+=1
    else: sp+=1
print("Uppercase =",u)
print("Lowercase =",l)
print("Digits =",d)
print("Special characters =",sp)
```

```python
s="programming"
ch="g"
first=last=-1
for i in range(len(s)):
    if s[i]==ch:
        if first==-1: first=i
        last=i
print("First occurrence =",first)
print("Last occurrence =",last)
```
