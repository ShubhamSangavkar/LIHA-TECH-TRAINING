## Section C - Advanced

**Q1.** Print Pascal's triangle for n rows using nested range() loops.

```python
rows = 5
for i in range(rows):
    num = 1
    for j in range(i + 1):
        print(num, end=" ")
        num = num * (i - j) // (j + 1)
    print()
```

**Q2.** Print a diamond star pattern using range().

```python
h = 4
for i in range(1, h + 1):
    print(" " * (h - i) + "*" * (2 * i - 1))
for i in range(h - 1, 0, -1):
    print(" " * (h - i) + "*" * (2 * i - 1))
```

**Q3.** Print all Armstrong numbers from 100 to 999 using range().

```python
for num in range(100, 1000):
    a = num // 100
    b = (num // 10) % 10
    c = num % 10
    if a ** 3 + b ** 3 + c ** 3 == num:
        print(num, end=" ")
```

**Q4.** Find all twin-prime pairs between 1 and 50 using range().

```python
primes = []
for num in range(2, 51):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(num)

for i in range(len(primes) - 1):
    if primes[i + 1] - primes[i] == 2:
        print((primes[i], primes[i + 1]), end=" ")
```

**Q5.** Transpose a matrix by iterating through row and column indices with range().

```python
matrix = [[1, 2, 3], [4, 5, 6]]
rows = len(matrix)
cols = len(matrix[0])
result = []
for j in range(cols):
    new_row = []
    for i in range(rows):
        new_row.append(matrix[i][j])
    result.append(new_row)
print(result)
```

**Q6.** Multiply two 2 x 2 matrices using three nested range() loops.

```python
A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]
result = [[0, 0], [0, 0]]
for i in range(2):
    for j in range(2):
        for k in range(2):
            result[i][j] = result[i][j] + A[i][k] * B[k][j]
print(result)
```

**Q7.** Sort a list using bubble sort implemented with nested range() loops.

```python
numbers = [5, 1, 4, 2, 8]
n = len(numbers)
for i in range(n - 1):
    for j in range(n - 1 - i):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
print(numbers)
```

**Q8.** Generate all prime numbers up to n using a sieve implemented with range().

```python
n = 30
is_prime = [True] * (n + 1)
is_prime[0] = False
is_prime[1] = False
for i in range(2, n + 1):
    if is_prime[i]:
        for j in range(i * 2, n + 1, i):
            is_prime[j] = False
for i in range(2, n + 1):
    if is_prime[i]:
        print(i, end=" ")
```

**Q9.** Display a hollow square star pattern using nested range() loops.

```python
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

**Q10.** Find all Pythagorean triples with values up to 20 using nested range() loops.

```python
limit = 20
for a in range(1, limit + 1):
    for b in range(a, limit + 1):
        for c in range(b, limit + 1):
            if a * a + b * b == c * c:
                print("(" + str(a) + "," + str(b) + "," + str(c) + ")", end=" ")
```
