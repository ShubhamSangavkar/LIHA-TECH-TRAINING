# Python Practice Sheet - List Comprehension

## Section A - Basic

**Q1.** Create a list of squares using list comprehension.

```python
numbers = [1, 2, 3, 4, 5]
squares = [n * n for n in numbers]
print(squares)
```

**Q2.** Extract all even numbers from a list.

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
evens = [n for n in numbers if n % 2 == 0]
print(evens)
```

**Q3.** Extract all odd numbers from a list.

```python
numbers = [1, 2, 3, 4, 5, 6, 7]
odds = [n for n in numbers if n % 2 != 0]
print(odds)
```

**Q4.** Convert all words in a list to uppercase.

```python
words = ['python', 'java', 'sql']
upper_words = [w.upper() for w in words]
print(upper_words)
```

**Q5.** Create a list containing the length of each word.

```python
words = ['AI', 'Python', 'LihaTech']
lengths = [len(w) for w in words]
print(lengths)
```

**Q6.** Select words having more than four characters.

```python
words = ['data', 'python', 'AI', 'model', 'cloud']
long_words = [w for w in words if len(w) > 4]
print(long_words)
```

**Q7.** Double every number in a list.

```python
numbers = [2, 4, 6, 8]
doubled = [n * 2 for n in numbers]
print(doubled)
```

**Q8.** Extract only positive numbers from a list.

```python
numbers = [-5, 3, -1, 0, 7, -2, 9]
positives = [n for n in numbers if n > 0]
print(positives)
```

**Q9.** Convert a string into a list of its characters.

```python
text = "Python"
chars = [ch for ch in text]
print(chars)
```

**Q10.** Extract only vowels from a string.

```python
text = "LihaTech Kolhapur"
vowels = [ch for ch in text if ch in "aeiouAEIOU"]
print(vowels)
```

**Q11.** Replace every negative number in a list with zero.

```python
numbers = [4, -2, 7, -5, 3]
result = [n if n >= 0 else 0 for n in numbers]
print(result)
```

**Q12.** Create a list containing the first character of each word.

```python
words = ['Python', 'Machine', 'Learning']
first_chars = [w[0] for w in words]
print(first_chars)
```

**Q13.** Create pairs of each number and its square.

```python
numbers = [1, 2, 3, 4]
pairs = [(n, n * n) for n in numbers]
print(pairs)
```

**Q14.** Flatten a two-dimensional list.

```python
matrix = [[1, 2], [3, 4], [5, 6]]
flat = [x for row in matrix for x in row]
print(flat)
```

**Q15.** Generate all multiples of 3 from 1 to 20.

```python
multiples = [n for n in range(1, 21) if n % 3 == 0]
print(multiples)
```

**Q16.** Convert a list of numeric strings into integers.

```python
values = ['10', '20', '30', '40']
numbers = [int(v) for v in values]
print(numbers)
```

## Section B - Intermediate

**Q1.** Label each number as Even or Odd using conditional list comprehension.

```python
numbers = [1, 2, 3, 4, 5]
labels = ['Even' if n % 2 == 0 else 'Odd' for n in numbers]
print(labels)
```

**Q2.** Transpose a matrix using nested list comprehension.

```python
matrix = [[1, 2, 3], [4, 5, 6]]
transpose = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(transpose)
```

**Q3.** Find common elements between two lists.

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
common = [x for x in list1 if x in list2]
print(common)
```

**Q4.** Remove empty or whitespace-only strings from a list.

```python
values = ['Python', '', ' ', 'AI', 'Data']
cleaned = [v for v in values if v.strip() != ""]
print(cleaned)
```

**Q5.** Flatten a nested list but keep only even numbers.

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8]]
evens = [x for row in matrix for x in row if x % 2 == 0]
print(evens)
```

**Q6.** Extract the names of students whose score is 75 or above.

```python
students = [('Asha', 82), ('Ravi', 68), ('Neha', 91), ('Aman', 75)]
names = [name for name, score in students if score >= 75]
print(names)
```

**Q7.** Generate the Cartesian product of two lists.

```python
list1 = [1, 2]
list2 = ['A', 'B']
product = [(a, b) for a in list1 for b in list2]
print(product)
```

**Q8.** Generate all prime numbers between 2 and 30.

```python
primes = [n for n in range(2, 31) if len([i for i in range(2, n) if n % i == 0]) == 0]
print(primes)
```

**Q9.** Extract all palindrome words from a list.

```python
words = ['level', 'python', 'madam', 'data', 'radar']
palindromes = [w for w in words if w == w[::-1]]
print(palindromes)
```

**Q10.** Extract the main diagonal elements of a square matrix.

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
diagonal = [matrix[i][i] for i in range(len(matrix))]
print(diagonal)
```

## Section C - Advanced

**Q1.** Find all unique elements that occur more than once while preserving their first appearance order.

```python
numbers = [4, 2, 4, 3, 2, 5, 4, 3]
repeated = [x for i, x in enumerate(numbers) if numbers.count(x) > 1 and numbers.index(x) == i]
print(repeated)
```

**Q2.** Generate all pairs whose sum is equal to a given target.

```python
numbers = [1, 2, 3, 4, 5, 6]
target = 7
pairs = [(a, b) for a in numbers for b in numbers if a < b and a + b == target]
print(pairs)
```

**Q3.** Create the transpose of a 3 x 3 matrix using nested list comprehension.

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
transpose = [[matrix[row][col] for row in range(3)] for col in range(3)]
print(transpose)
```

**Q4.** Flatten a three-level nested list into a single list.

```python
data = [[[1, 2], [3]], [[4, 5], [6]]]
flat = [x for level1 in data for level2 in level1 for x in level2]
print(flat)
```

**Q5.** Calculate the sum of each row of a matrix.

```python
matrix = [[2, 4, 6], [1, 3, 5], [10, 20, 30]]
row_sums = [sum(row) for row in matrix]
print(row_sums)
```

**Q6.** Assign grades to marks: A for 90+, B for 75-89, C for 60-74, and D below 60.

```python
marks = [95, 82, 67, 54, 90]
grades = ['A' if m >= 90 else 'B' if m >= 75 else 'C' if m >= 60 else 'D' for m in marks]
print(grades)
```

**Q7.** Convert comma-separated student records into tuples.

```python
records = ['Asha,85', 'Ravi,72', 'Neha,91']
students = [(r.split(',')[0], int(r.split(',')[1])) for r in records]
print(students)
```

**Q8.** Create a multiplication table from 1 to 5 as a nested list.

```python
table = [[i * j for j in range(1, 6)] for i in range(1, 6)]
print(table)
```

**Q9.** Generate all unique Pythagorean triples where 1 <= a < b < c <= 20.

```python
triples = [(a, b, c) for a in range(1, 21) for b in range(a + 1, 21) for c in range(b + 1, 21) if a * a + b * b == c * c]
print(triples)
```

**Q10.** Generate sliding windows of size 3 from a list.

```python
numbers = [10, 20, 30, 40, 50, 60]
windows = [numbers[i:i + 3] for i in range(len(numbers) - 2)]
print(windows)
```
