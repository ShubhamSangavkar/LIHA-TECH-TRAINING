**Q1.** Compose two functions so that the output of the first becomes the input of the second.

```python
def add_three(x):
    return x + 3

def square(x):
    return x * x

def compose(first, second):
    def combined(x):
        return second(first(x))
    return combined

func = compose(add_three, square)
print(func(5))
```

**Q2.** Build a transformation pipeline that strips spaces, converts text to lowercase, and replaces spaces with hyphens.

**Q6.** Function that returns a counter function. Each call to the returned function should increase the count.

```python
def make_counter():
    count = 0

    def counter():
        nonlocal count
        count += 1
        return count

    return counter

counter = make_counter()
results = []
for i in range(4):
    results.append(counter())
print(results)
```

**Q9.** Apply a list of functions to the same input value.

```python
def square(x):
    return x * x

def cube(x):
    return x ** 3

def double(x):
    return x * 2

def apply_all(functions, value):
    results = []
    for f in functions:
        results.append(f(value))
    return results

print(apply_all([square, cube, double], 3))
```

**Q1.** Curried function where adding three values is performed through three successive function calls.

```python
def add(a):
    def add_b(b):
        def add_c(c):
            return a + b + c
        return add_c
    return add_b

print(add(10)(20)(30))
```

**Q2.** Function that accepts a list of functions and builds a reusable processing pipeline.

```python
def double(x):
    return x * 2

def add_one(x):
    return x + 1

def square(x):
    return x * x

def build_pipeline(functions):
    def run(value):
        for f in functions:
            value = f(value)
        return value
    return run

pipeline = build_pipeline([double, add_one, square])
print(pipeline(3))
```

**Q3.** Write your own higher-order my_map() function without using Python's built-in map().

```python
def square(x):
    return x * x

def my_map(func, data):
    result = []
    for item in data:
        result.append(func(item))
    return result

print(my_map(square, [1, 2, 3, 4]))
```

**Q4.** Write your own higher-order my_filter() function without using Python's built-in filter().

```python
def is_even(x):
    return x % 2 == 0

def my_filter(predicate, data):
    result = []
    for item in data:
        if predicate(item):
            result.append(item)
    return result

print(my_filter(is_even, [1, 2, 3, 4, 5, 6]))
```

**Q5.** Write your own higher-order my_reduce() function without using functools.reduce().

```python
def add(a, b):
    return a + b

def my_reduce(func, data):
    result = data[0]
    for item in data[1:]:
        result = func(result, item)
    return result

print(my_reduce(add, [10, 20, 30, 40]))
```

**Q6.** Higher-order memoize function that wraps a function and caches previously calculated results.

```python
calculations = 0

def square(x):
    global calculations
    calculations += 1
    return x * x

def memoize(func):
    cache = {}

    def wrapper(x):
        if x not in cache:
            cache[x] = func(x)
        return cache[x]

    return wrapper

fast_square = memoize(square)
print("Results =", fast_square(5), fast_square(5), fast_square(7))
print("Actual calculations =", calculations)
```

**Q7.** Higher-order retry function that calls another function again when it fails, up to a given retry limit.

```python
attempts = 0

def unstable():
    global attempts
    attempts += 1
    if attempts < 3:
        raise Exception("Failed")
    return "Done"

def retry(func, retries):
    for attempt in range(1, retries + 1):
        try:
            func()
            return attempt
        except Exception:
            pass
    return None

result = retry(unstable, 3)
if result is not None:
    print("Success on attempt", result)
else:
    print("All attempts failed")
```

**Q8.** Function factory that creates different discount policies and returns the appropriate pricing function.

```python
def make_pricing(policy):
    if policy == 'premium':
        discount = 15
    elif policy == 'regular':
        discount = 5
    else:
        discount = 0

    def pricing(price):
        return price - price * discount / 100

    return pricing

price_func = make_pricing('premium')
print("Final price =", price_func(2000))
```

**Q10.** Strategy-based higher-order function that selects a processing function at runtime according to a mode.

```python
def normalize(data):
    low = min(data)
    high = max(data)
    return [(x - low) / (high - low) for x in data]

def double_all(data):
    return [x * 2 for x in data]

def process(mode, data):
    strategies = {'normalize': normalize, 'double': double_all}
    return strategies[mode](data)

print(process('normalize', [10, 20, 30]))
```
