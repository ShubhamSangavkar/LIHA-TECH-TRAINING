**Q8.** Count the number of vowels in a string.

```python
def count_vowels(text):
    count = 0
    for ch in text.lower():
        if ch in "aeiou":
            count += 1
    return count

print("Vowel count =", count_vowels("Programming"))
```

**Q10.** Calculate the factorial of a number.

```python
def factorial(number):
    result = 1
    for i in range(1, number + 1):
        result = result * i
    return result

print("Factorial =", factorial(5))
```

**Q1.** Return the minimum, maximum, and average of a list in a single call.

```python
def list_stats(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)

minimum, maximum, avg = list_stats([12, 5, 30, 18, 9])
print("Minimum =", minimum)
print("Maximum =", maximum)
print("Average =", avg)
```

**Q2.** Use *args to return the sum of any number of arguments.

```python
def total(*args):
    result = 0
    for value in args:
        result += value
    return result

print("Sum =", total(10, 20, 30, 40, 50))
```

**Q3.** Use **kwargs to format employee information.

```python
def employee_info(**kwargs):
    parts = []
    for key, value in kwargs.items():
        parts.append(key.capitalize() + ": " + str(value))
    return ", ".join(parts)

print(employee_info(name='Asha', role='Developer', city='Pune'))
```

**Q4.** Function that updates and returns a global counter every time it is called.

```python
counter = 0

def increment():
    global counter
    counter += 1
    return counter

print("Counter values =", increment(), increment(), increment())
```

**Q5.** Count the frequency of each element in a list and return a dictionary.

```python
def count_frequency(items):
    freq = {}
    for item in items:
        if item in freq:
            freq[item] += 1
        else:
            freq[item] = 1
    return freq

print(count_frequency(['A', 'B', 'A', 'C', 'B', 'A']))
```

**Q6.** Recursive function to find the sum of digits of a number.

```python
def digit_sum(number):
    if number == 0:
        return 0
    return number % 10 + digit_sum(number // 10)

print("Sum of digits =", digit_sum(5832))
```

**Q7.** Recursive function to return the nth Fibonacci number.

```python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print("Fibonacci number =", fibonacci(8))
```

**Q8.** Calculate income tax using slabs: 0% up to 300000, 5% for the next 300000, and 10% for the remaining amount.

```python
def income_tax(income):
    tax = 0
    if income > 600000:
        tax = tax + (income - 600000) * 0.10
        tax = tax + 300000 * 0.05
    elif income > 300000:
        tax = tax + (income - 300000) * 0.05
    return tax

print("Tax =", income_tax(800000))
```

**Q9.** Function containing a nested helper function to calculate gross salary.

```python
def gross_salary(basic, hra, da):
    def percent(value, p):
        return value * p / 100

    return basic + percent(basic, hra) + percent(basic, da)

print("Gross salary =", gross_salary(30000, 20, 10))
```

**Q10.** Validate a password and return both status and reason.

```python
def validate_password(password):
    if len(password) < 8:
        return False, "Password must have at least 8 characters"

    has_upper = False
    has_digit = False
    for ch in password:
        if ch.isupper():
            has_upper = True
        if ch.isdigit():
            has_digit = True

    if not has_upper:
        return False, "Password must have an uppercase letter"
    if not has_digit:
        return False, "Password must have a digit"
    return True, "Password accepted"

status, reason = validate_password("Python123")
print("Valid =", status)
print("Reason =", reason)
```

**Q1.** Recursive function to flatten a nested list containing values at different nesting levels.

```python
def flatten(data):
    result = []
    for item in data:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

print(flatten([1, [2, [3, 4], 5], [6, 7]]))
```

**Q2.** Binary search on a sorted list returning the index of the target value.

```python
def binary_search(numbers, target):
    low = 0
    high = len(numbers) - 1
    while low <= high:
        mid = (low + high) // 2
        if numbers[mid] == target:
            return mid
        elif numbers[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

print("Index =", binary_search([2, 5, 8, 12, 16, 23, 38], 16))
```

**Q3.** Merge overlapping intervals.

```python
def merge_intervals(intervals):
    intervals = sorted(intervals)
    merged = []
    for start, end in intervals:
        if len(merged) == 0 or merged[-1][1] < start:
            merged.append((start, end))
        else:
            merged[-1] = (merged[-1][0], max(merged[-1][1], end))
    return merged

print(merge_intervals([(1, 3), (2, 6), (8, 10), (9, 12)]))
```

**Q4.** Find the longest common prefix among a list of strings.

```python
def common_prefix(words):
    prefix = words[0]
    for word in words[1:]:
        while not word.startswith(prefix):
            prefix = prefix[:-1]
    return prefix

print("Longest common prefix =", common_prefix(['flower', 'flow', 'flight']))
```

**Q5.** Group anagrams together.

```python
def group_anagrams(words):
    groups = {}
    for word in words:
        key = "".join(sorted(word))
        if key not in groups:
            groups[key] = []
        groups[key].append(word)
    return list(groups.values())

print(group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']))
```

**Q6.** Return the k most frequent elements in a list.

```python
def top_k_frequent(numbers, k):
    counts = {}
    for n in numbers:
        counts[n] = counts.get(n, 0) + 1
    ordered = sorted(counts, key=counts.get, reverse=True)
    return ordered[:k]

print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))
```

**Q7.** Generate the nth row of Pascal's Triangle.

```python
def pascal_row(n):
    row = [1]
    for i in range(n - 1):
        new_row = [1]
        for j in range(len(row) - 1):
            new_row.append(row[j] + row[j + 1])
        new_row.append(1)
        row = new_row
    return row

print(pascal_row(5))
```

**Q8.** Simple calculator based on an operation argument.

```python
def calculator(a, b, operation):
    if operation == 'add':
        return a + b
    elif operation == 'subtract':
        return a - b
    elif operation == 'multiply':
        return a * b
    elif operation == 'divide':
        if b == 0:
            return "Cannot divide by zero"
        return a / b
    return "Invalid operation"

print("Result =", calculator(20, 5, 'divide'))
```

**Q9.** Multiply two matrices and return the resulting matrix.

```python
def multiply_matrices(A, B):
    rows = len(A)
    cols = len(B[0])
    inner = len(B)
    result = []
    for i in range(rows):
        row = []
        for j in range(cols):
            total = 0
            for k in range(inner):
                total += A[i][k] * B[k][j]
            row.append(total)
        result.append(row)
    return result

print(multiply_matrices([[1, 2], [3, 4]], [[5, 6], [7, 8]]))
```

**Q10.** Validate whether brackets in an expression are balanced.

```python
def is_balanced(expression):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    for ch in expression:
        if ch in "([{":
            stack.append(ch)
        elif ch in ")]}":
            if len(stack) == 0 or stack.pop() != pairs[ch]:
                return False
    return len(stack) == 0

print("Balanced =", is_balanced("{[()]}"))
```
