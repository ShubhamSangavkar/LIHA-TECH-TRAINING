# **MATPLOTLIB - DATA VISUALIATION**

```python
import matplotlib.pyplot as plt
plt
```

LINE PLOT

```python
#LiNE plot is univariate , BI variate and multi as well

days= ['mon','tue','wed','thur','fri']
temp = [25,31,34,27,29]
```

```python
plt.plot(days,temp)
```

```python
plt.plot(days,temp,"m--")
```

```python

# plt.plot(days,temp,"o")
# plt.plot(days,temp,"m")

plt.plot(days,temp,"k--*")
```

```python

plt.figure(figsize=(4,4))
plt.plot(days,temp,"m--")
plt.plot(days,temp,"or",label = 'Temp.trend')
plt.xlabel("days")
plt.ylabel("temperatture")
plt.legend()
plt.title("Temperature Analysys")
plt.savefig("temp.png",dpi = 350)
```

```python
state = ['Maharastra', 'Goa',"Rajastan","Uttar Pradesh","Himachal Pradesh"]
count = [1000,350,900,1500,2000]
```

```python
plt.plot(state,count,"r--")
plt.plot(state,count,"ok",label = "")
plt.xlabel("State")
plt.ylabel("Count")
plt.title("State Analysis")
```

```python
plt.bar(state,count)
plt.xlabel("State")
plt.ylabel("Count")
plt.title("State Analysis")
```

```python
plt.barh(state,count,color = "r")
plt.xlabel("State")
plt.ylabel("Count")
plt.title("State Analysis")
```

```python
plt.pie(count,autopct="%.2f%%",labels=state)
plt.show()
```

```python
gen = ["male","female"]
count = [2300,5000]

plt.pie(count,autopct="%.2f%%",labels=gen)
plt.show()
```
