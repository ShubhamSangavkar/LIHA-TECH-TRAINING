```python
class Oprstn:
    def add(self,a,b):
        print("addition :" , a+b)
    def add(self,a,b,c):
        print("addition :", a+b+c)
    def add(self):
        print("zero")

ops=Oprstn()
ops.add()
ops.add(10,20)
```

```python
class Oprstn:
    def add(self,*args,**kwargs):
        print("addition of args:",sum(args))
        print("addition of kwargs : ",sum(kwargs.values()))
        print("--"*20)
ops=Oprstn()

ops.add(10,20,20,a=200,b=300,c=400)
ops.add(10,20,30)
ops.add()
ops.add(a=100,b=300)
```

```python
try :
    print(2/0)
except ZeroDivisionError as msg:
    print("ERROR : ",msg)
else:
    print("No ERROR")
finally:
    print("EXECUTION COMPLETED")
```

```python
try :
    balance=10000
    
    withdraw = int(input("Enter Amount to Withdraw"))
    if withdraw > balance:
        raise ValueError("Insufficient Balance")
except ValueError as msg:
    print("Error : ", msg)
else:
    balance -= withdraw
    print("Transaction Successful. Remaining Balance : " ,balance)
finally:
    print("THANK YOU")
```
